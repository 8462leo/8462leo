```python
#Jogo do numero aleatorio, tente acertar o numero com poucas tentativas.

from contextlib import nullcontext
import random

aleatorio = random.randint(1,100)
chute = nullcontext
tentativas = 1

while chute != aleatorio:
  chute = int(input("adivinhe o numero entre 1 a 100: "))
  print("")
  if chute > aleatorio:
    print(f"O numero é menor que {chute}!")
    tentativas += 1
  elif chute < aleatorio:
    print(f"O numero é maior que {chute}!")
    tentativas += 1
  else:
    print(f"Você acertou o numero aleatorio {aleatorio} em {tentativas} tentativas.")
    break
```

```python
from ctypes import c_char
# Jogo da forca!

import random

print("")
print("--- Bem vindo ao jogo da forca! ---")
print("")

print("Dica, a palavra secreta é uma linguagem de programação!")

palavras_base = ["Python", "Java", "JavaScript", "React", "PHP", "Ruby", "TypeScript", "Swift", "Kotlin", "Dart", "Assembly"] # Lista de possiveis palavras no jogo
palavra_secreta = random.choice(palavras_base) # Escolhe uma palavra aleatoria
lower_palavra_secreta = palavra_secreta.lower() # Deixa todas as letras minusculas

tamanho_palavra = len(palavra_secreta) # Conta a quantidade de letras da palavra secreta

print("")
print(f"A palavra secreta tem {tamanho_palavra} letras.")

acertos = []
erros = 0
chances = 10

# Display initial hidden word
for letra in lower_palavra_secreta:
  print(f"_", end=" ") # Adiciona os _ comforme a palavra
print("")

while erros < chances and set(acertos) != set(lower_palavra_secreta):

    print("")
    chute = input("Digite uma letra:").lower() # Chute do usuario e conversão para lowcase
    while True:
      if len(chute) == 1 and chute.isalpha():
        break
      else:
        chute = input("Digite apenas uma letra válida: ").lower()

    if chute in acertos:
        print(f"Você já tentou a letra '{chute}'.") #Validação da letra

        for letra in lower_palavra_secreta:
            if letra in acertos:
                print(f"{letra} ", end="")
            else:
                print("_ ", end="")
        print("")
        continue

    elif chute in lower_palavra_secreta: # Se o chute estiver na palavra secreta adiciona a letra no acerto.
      acertos.append(chute)
      print(f"Acertou a letra '{chute}'!")

      for letra in lower_palavra_secreta:
        if letra in acertos:
          print(f"{letra} ", end="")
        else:
          print("_ ", end="")
      print("")

    else:
          erros += 1
          print(f"A letra '{chute}' não está na palavra.")
          print(f"Você tem {chances - erros} chances restantes.")

          for letra in lower_palavra_secreta: #Repete o codigo
              if letra in acertos:
                  print(f"{letra} ", end="")
              else:
                  print("_ ", end="")
          print("")


if set(acertos) == set(lower_palavra_secreta): # Verifica se vc ganhou ou perdeu após o loop
    print("")
    print(f"Parabéns! Você acertou a palavra {palavra_secreta}!")
else:
    print("")
    print(f"Você perdeu! A palavra secreta era '{palavra_secreta}'.")
```

```python
# Previsão do tempo

import requests
from google.colab import userdata

# Chave API KEY

api_key = userdata.get('previsao_tempo')

local = input("Qual o local que deseja saber a previsão do tempo? ")

url_base = f"http://api.openweathermap.org/data/2.5/weather?"

# Monta a URL completa com os parâmetros necessários
url_completa = f"{url_base}appid={api_key}&q={local}&lang=pt_br"

try:
  # Converte a resposta json em python
  dados = requests.get(url_completa).json()

  # Verifica se a requisição deu certo
  if dados ["cod"] != "404":

    # Extraindo dados
    temperatura = dados ["main"] ["temp"] - 273.15
    descricao = dados ["weather"] [0] ["description"]
    umidade = dados ["main"] ["humidity"]

    print("")
    print(f" A descrição do tempo é {descricao}")
    print(f" A temperatura atual é {temperatura: .2f}°C")
    print(f" A umidade é de {umidade}%")

  else:
    print("Cidade não encontrada!")

except:
  print("Erro na requisição!")
```

```python
# Programa de últimas notícias

import requests
from google.colab import userdata

newsapi = userdata.get('ultimas_noticias')
chave_api = newsapi

assunto = input("Qual assunto deseja pesquisar? ")
categoria = input("Qual categoria deseja pesquisar? ")

# Cria a URL de forma mais segura e correta, usando 'params'
url_base = "https://newsapi.org/v2/everything?"

params = {
    'q': assunto,
    'language': 'pt',
    'category': categoria,
    'country': 'br',
    'apiKey': chave_api
}

try:
    # Faz a requisição para a API
    response = requests.get(url_base, params=params)
    response.raise_for_status() # Verifica se a requisição foi bem-sucedida
    dados = response.json()

    # Verifica se a lista de artigos não está vazia
    if dados["articles"]:
        primeira_noticia = dados["articles"][0]

        titulo = primeira_noticia["title"]
        descricao = primeira_noticia["description"]
        content = primeira_noticia["content"]

        print("\n---")
        print(f"Título: {titulo}")
        print(f"Descrição: {descricao}")
        print(f"Conteúdo: {content}")
    else:
        print("\nNenhuma notícia encontrada com os critérios de pesquisa.")

except requests.exceptions.RequestException as e:
    print(f"\nOcorreu um erro na requisição: {e}")
```
